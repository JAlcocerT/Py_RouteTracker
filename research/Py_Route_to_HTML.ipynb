#### Defining the functions for multi-route

```python
# import gpxpy
# import pandas as pd
# import folium

def plot_routes_on_map(gpx_paths, colors):
    route_data = []
    for gpx_path, color in zip(gpx_paths, colors):
        with open(gpx_path, 'r') as gpx_file:
            gpx = gpxpy.parse(gpx_file)
            for track in gpx.tracks:
                for segment in track.segments:
                    for point in segment.points:
                        route_data.append({
                            'latitude': point.latitude,
                            'longitude': point.longitude,
                            'elevation': point.elevation,
                            'color': color
                        })

    route_df = pd.DataFrame(route_data)

    route_map = folium.Map(
        location=[50, 23],
        zoom_start=7,
        tiles='OpenStreetMap',
        #tiles='Stamen Toner',
        #tiles="Stamen Terrain",
        #tiles="Stamen Watercolor",
        width=924,
        height=600
    )

    for color, route_df_group in route_df.groupby('color'):
        coordinates = [tuple(x) for x in route_df_group[['latitude', 'longitude']].to_numpy()]
        folium.PolyLine(coordinates, weight=3, color=color).add_to(route_map)

    folium.LayerControl().add_to(route_map)

    return route_map
```

```python
route_map = plot_routes_on_map(['My_Routes/Krakow - Zarki.gpx',
                                'My_Routes/Zarki - Czestochowa.gpx'], ['#027835',
                                                                       '#027835'])
display(route_map)
```

```python
def plot_routes_on_map_markers(gpx_paths, colors, markers=[]):
    route_data = []
    for gpx_path, color in zip(gpx_paths, colors):
        with open(gpx_path, 'r') as gpx_file:
            gpx = gpxpy.parse(gpx_file)
            for track in gpx.tracks:
                for segment in track.segments:
                    for point in segment.points:
                        route_data.append({
                            'latitude': point.latitude,
                            'longitude': point.longitude,
                            'elevation': point.elevation,
                            'color': color
                        })

    route_df = pd.DataFrame(route_data)

    route_map = folium.Map(
        location=[47, 32],
        zoom_start=5,
        tiles='OpenStreetMap',
        width=924,
        height=600
    )

    for color, route_df_group in route_df.groupby('color'):
        coordinates = [tuple(x) for x in route_df_group[['latitude', 'longitude']].to_numpy()]
        folium.PolyLine(coordinates, weight=3, color=color).add_to(route_map)

    for marker in markers:
        folium.Marker(location=[marker['lat'], marker['lon']], popup=marker['name']).add_to(route_map)

    folium.LayerControl().add_to(route_map)

    return route_map
```

```python
markers = [{'lat': 50, 'lon': 20, 'name': 'Marker 1'},
           {'lat': 55, 'lon': 25, 'name': 'Marker 2'}]

route_map = plot_routes_on_map_markers(['My_Routes/PL-SK-HU.gpx', 'HU-GR_gpx/HU-GR.gpx'],
                                       ['green', 'gray'],
                                       markers=markers)
display(route_map)
```
